```python
'''
La funció encripta un text, inputValue, substituint digits del text per caracters alfanumerics, diccionari keys, 
utilitzant caràcters d'encriptació, tokens.
En el cas que no assignem valors a inputvalue, keys o token, la funció ens preguntarà per valors mitjançant un "input"
(inputValue = '',keys = '',token = ''))

'''
def encrypt(inputValue,keys,token): 
    
    if keys == '':
    #Creem un diccionari amb lletres i el seu equivalent
        keys = {"a" : 5, "e" : 4,"i" : 3,"o" : 2,"u" : 1,
        "A" : 55, "E" : 44,"I" : 33,"O" : 22,"U" : 11,
        "c" : 95, "m" : 94,"n" : 93,"l" : 92,"s" : 91,
        "d" : 85, "b" : 84,"g" : 83,"j" : 82,"t" : 81,
        "C" : 75, "M" : 74,"N" : 73,"L" : 72,"S" : 71,
        "D" : 65, "B" : 64,"G" : 63,"J" : 62,"T" : 61,
        "v" : 20, "h" : 30,"p" : 40,"q" : 50,
        " " : " ","f" : 99,
        "r" : 23 ,"V":24,"H":25,"P":26,"Q":27,"R":67 }
    #Creem un tupla amb els keys no admesos i pels quals s'haurà de donar un missatge d'error
    token_tupple = ("=","*","+","#","&","%")
    #Es demana el text a encriptar
    if token == '':
        token = input("Diga'm quin token vols utilitzar. Token disponibles = * + # & %")
    #verificar si el token escollit està dintre de la tupla
    if token not in token_tupple:
        #en cas de no ser-hi, imprimeix aquests missatge.
        return ("Glups! El token escollit no és vàlid. Si us plau, escolleixi un nou token")
    if inputValue == '':
        inputValue = input("Introdueix un text a encriptar")
    missatge = ''
    missatge_definitiu = ""
    a = 0
    #Booleà que s'inicialitza la vatiable "available_token" com False i en cas d'haver-hi un token de la tupla se li assignarà el valor de True.
    available_token = False
    #Itinarem per cada lletra del inputValue per verficar si hi ha un token de la tupla
    for char in inputValue:
        #preguntem si hi ha token al caracter
        if char in token:
            #En cas afirmatiu, assigna un valor True a la variable available_token
            available_token = True          
            #surt del bucle
            break
    #pregunta si la variable és igual a True or False. En case de True, imprimeix els missatge i s'acaba.
    if available_token == True:
        return ("No es posible encriptar la cadena. Verifique los parámetros de entrada, esto es debido a que la cadena de entrada inputValue, tiene incluido en su texto caracteres iguales al parámetro token")     
    # comença bucle encriptació
    else:
    #Revisa un per un els caràcters del text
        for char in inputValue:
            #obtenim el valor del caràcter
            missatge = (keys.get(char))
            #està el caràcter al diccionari?
            if char in keys.keys():
                #Si la primera lletra està en el inputValue afageix el token
                if a == 0:
                #afageix la variable al missatge_definitiu més els tokens corresponents
                    missatge_definitiu += token + str(missatge) + token
                    a+=1
                # només afeix el token al final, per evitar tockens dobles en el cas que hi hagin dos digits seguits que estiguin en el diccionari.
                else:
                    missatge_definitiu += (str(missatge) + token)
            #el caràcter no està al diccionari
            else:
                #afageix la variable al missatge definitiu
                missatge_definitiu += char
    #retorna el missatge_definitu
    return missatge_definitiu

print(encrypt(inputValue = 'Opcion que permite generar una solicitud',keys = '',token = '*'))
    
```
